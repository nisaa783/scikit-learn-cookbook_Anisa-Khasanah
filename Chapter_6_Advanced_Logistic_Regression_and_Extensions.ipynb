## **1. Pendahuluan dan Konsep Dasar Logistic Regression**
Meskipun menyandang kata "Regresi" pada namanya, Logistic Regression adalah salah satu algoritma machine learning paling populer yang digunakan untuk tugas klasifikasi biner (memprediksi dua kelas, misalnya 0 atau 1, ya atau tidak). Model ini menghitung probabilitas bahwa suatu sampel data termasuk ke dalam kelas tertentu menggunakan fungsi logistik atau fungsi sigmoid.Dalam praktiknya, masalah klasifikasi sering kali melibatkan lebih dari dua kategori (multikelas). Bab ini membahas bagaimana memperluas regresi logistik dasar ke dalam ranah multikelas serta menerapkan regularisasi untuk menjaga kestabilan model.

## **2. Fungsi Sigmoid dan Multiclass Classification**
Untuk memetakan nilai keluaran linier ke dalam rentang probabilitas antara 0 dan 1, regresi logistik menggunakan Fungsi Sigmoid.Untuk menangani klasifikasi multikelas (lebih dari dua kelas), scikit-learn mendukung dua strategi utama:Multinomial (Softmax): Mengoptimalkan fungsi kerugian multinomial secara langsung untuk memperkirakan probabilitas semua kelas sekaligus.One-vs-Rest (OvR): Memecah masalah multikelas menjadi beberapa tugas klasifikasi biner terpisah untuk setiap kelas terhadap kelas lainnya.Konsep & Rumus Matematika:Fungsi Sigmoid:$$\sigma(z) = \frac{1}{1 + e^{-z}}$$di mana $z = w^T x + b$ adalah kombinasi linier dari fitur dan bobot model.Binary Cross-Entropy Loss (dengan penalti $L_2$):$$J(w) = -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} \log(\hat{y}^{(i)}) + (1-y^{(i)})\log(1-\hat{y}^{(i)}) \right] + \frac{\lambda}{2m} \vert{}\vert{}w\vert{}\vert{}^2$$

## **3. Contoh Program Penerapan 1: Klasifikasi Multikelas Menggunakan Logistic Regression**
Contoh program ini memperlihatkan cara melatih model regresi logistik untuk menangani klasifikasi multikelas pada dataset Iris (yang memiliki 3 kelas bunga) menggunakan pendekatan multinomial.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# 1. Memuat dataset Iris (memiliki 3 kelas target: 0, 1, 2)
iris = load_iris()
X, y = iris.data, iris.target

# Membagi dataset menjadi data latih dan data uji
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 2. Inisialisasi dan melatih model Logistic Regression untuk multikelas (Multinomial)
# Parameter multi_class='multinomial' menggunakan fungsi Softmax
log_reg = LogisticRegression(multi_class='multinomial', solver='lbfgs', max_iter=200, random_state=42)
log_reg.fit(X_train, y_train)

# 3. Melakukan prediksi dan evaluasi model
y_pred = log_reg.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"Akurasi Logistic Regression Multiclass: {accuracy:.4f}")
print("\nLaporan Klasifikasi Lengkap:")
print(classification_report(y_test, y_pred, target_names=iris.target_names))
```

## **4. Contoh Program Penerapan 2: Logistic Regression dengan Regularisasi**
Sama seperti model linear lainnya, regresi logistik dapat mengalami overfitting jika jumlah fitur sangat besar. Kita dapat menerapkan penalti regularisasi ($L_1$ atau $L_2$) dengan mengatur parameter kekuatan invers regulasi C (di mana nilai C yang lebih kecil menunjukkan penalti regularisasi yang lebih kuat).

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

# 1. Memuat dataset Breast Cancer (klasifikasi biner)
data = load_breast_cancer()
X, y = data.data, data.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# 2. Inisialisasi Logistic Regression dengan regularisasi L2 (penalty default='l2')
# Parameter C=0.5 mengontrol kekuatan regularisasi
log_reg_reg = LogisticRegression(penalty='l2', C=0.5, solver='liblinear', random_state=42)
log_reg_reg.fit(X_train, y_train)

# 3. Evaluasi performa model reguler
train_acc = log_reg_reg.score(X_train, y_train)
test_acc = log_reg_reg.score(X_test, y_test)

print(f"Akurasi pada Data Latih: {train_acc:.4f}")
print(f"Akurasi pada Data Uji: {test_acc:.4f}")

# Menampilkan Confusion Matrix
y_pred = log_reg_reg.predict(X_test)
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```
