## 1. Pendahuluan dan Konsep Dasar Support Vector Machines (SVM)
Support Vector Machines (SVM) adalah salah satu algoritma machine learning paling tangguh dan fleksibel untuk tugas klasifikasi maupun regresi. Prinsip kerja utama SVM adalah mencari garis atau bidang pemisah optimal (hyperplane) yang memisahkan kelas data yang berbeda dengan margin maksimum (jarak terjauh antara hyperplane dengan titik data terdekat dari masing-masing kelas). Titik-titik data yang berada paling dekat dengan hyperplane dan menentukan lebar margin tersebut dikenal sebagai Support Vectors.

## 2. Kernel Trick dan Penanganan Data Non-Linear
Di dunia nyata, data sering kali tidak dapat dipisahkan secara linier di ruang fitur aslinya (misalnya data yang tersusun melingkar atau bergelombang). Untuk mengatasi masalah ini, SVM menggunakan teknik yang disebut Kernel Trick.Alih-alih mentransformasikan data secara eksplisit ke dimensi yang jauh lebih tinggi (yang membutuhkan komputasi sangat berat), fungsi kernel menghitung produk titik (dot product) langsung di ruang berdimensi tinggi tersebut. Salah satu kernel yang paling populer dan ampuh adalah Radial Basis Function (RBF) / Gaussian Kernel.
Konsep & Rumus Matematika:
Fungsi Keputusan SVM (Decision Function):$$f(x) = \sum_{i=1}^{n} \alpha_i y_i K(x_i, x) + b$$
Kernel RBF (Gaussian Kernel)
:$$K(x, x') = \exp(-\gamma \vert{}\vert{}x - x'\vert{}\vert{}^2)$$di mana $\gamma$ (gamma) menentukan seberapa jauh pengaruh satu titik data pelatihan (nilai $\gamma$ besar berarti pengaruhnya sempit dan lokal, sedangkan nilai kecil berarti pengaruhnya luas).

## 3. Contoh Program Penerapan 1: Klasifikasi Non-Linear Menggunakan SVM Kernel RBF
Contoh program ini memperlihatkan cara membuat dataset non-linear menggunakan fungsi make_moons dan melatih model SVC dengan kernel RBF untuk memisahkan data tersebut secara akurat.

```python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# 1. Membuat dataset non-linear berbentuk bulan sabit (moons)
X, y = make_moons(n_samples=300, noise=0.2, random_state=42)

# Membagi dataset menjadi data latih dan data uji
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 2. Inisialisasi dan melatih SVM dengan Kernel RBF
# Parameter C mengontrol trade-off antara margin lebar dan kesalahan klasifikasi
# Parameter gamma mengontrol seberapa kompleks batas keputusan (decision boundary)
svm_model = SVC(kernel='rbf', C=1.0, gamma='scale', random_state=42)
svm_model.fit(X_train, y_train)

# 3. Melakukan prediksi dan evaluasi model
y_pred = svm_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"Akurasi SVM Kernel RBF: {accuracy:.4f}")
print(f"Jumlah Support Vectors yang dihasilkan: {svm_model.n_support_}")
print("\nLaporan Klasifikasi Lengkap:")
print(classification_report(y_test, y_pred))
```

## 4. Contoh Program Penerapan 2: Hyperparameter Tuning SVM Menggunakan GridSearchCV
Performa model SVM sangat sensitif terhadap kombinasi parameter $C$ (regularisasi) dan $\gamma$ (kernel coefficient). Kita dapat menggunakan GridSearchCV untuk menemukan kombinasi parameter terbaik secara otomatis melalui validasi silang.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.svm import SVC

# 1. Memuat dataset Breast Cancer
data = load_breast_cancer()
X, y = data.data, data.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# 2. Menentukan rentang parameter C dan gamma yang akan diuji
param_grid = {
    'C': [0.1, 1, 10, 100],
    'gamma': ['scale', 'auto', 0.01, 0.1, 1],
    'kernel': ['rbf']
}

# 3. Inisialisasi GridSearchCV dengan 5-fold cross-validation
grid_search = GridSearchCV(estimator=SVC(), param_grid=param_grid, cv=5, scoring='accuracy')
grid_search.fit(X_train, y_train)

# 4. Menampilkan hasil parameter terbaik dan evaluasi pada data uji
print(f"Parameter Terbaik yang Ditemukan: {grid_search.best_params_}")
print(f"Akurasi Terbaik pada Cross-Validation: {grid_search.best_score_:.4f}")

best_svm = grid_search.best_estimator_
test_accuracy = best_svm.score(X_test, y_test)
print(f"Akurasi Model Terbaik pada Data Uji: {test_accuracy:.4f}")
```
