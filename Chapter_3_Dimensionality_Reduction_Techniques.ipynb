## **1. Pendahuluan dan Pentingnya Reduksi Dimensi**
Ketika kita bekerja dengan dataset yang memiliki puluhan, ratusan, atau bahkan ribuan fitur, model machine learning sering kali menghadapi fenomena yang dikenal sebagai Curse of Dimensionality (Kutukan Dimensi). Semakin tinggi dimensi data, volume ruang fitur menjadi sangat jarang (sparse), membuat jarak antar titik data kehilangan makna statistik, memicu overfitting, dan memperlambat waktu komputasi secara drastis.Reduksi dimensi bertujuan untuk memangkas jumlah fitur yang tidak penting atau berlebihan sambil mempertahankan informasi atau varians esensial sebanyak mungkin. Teknik ini terbagi menjadi dua pendekatan utama: unsupervised (seperti PCA) dan supervised (seperti LDA).

## **2. Principal Component Analysis (PCA)**
PCA adalah teknik unsupervised learning yang paling populer untuk mereduksi dimensi. PCA bekerja dengan mencari arah atau sumbu ortogonal baru—yang disebut sebagai Principal Components (Komponen Utama)—di mana varians data tersebar paling maksimal. Sumbu-sumbu baru ini merupakan kombinasi linier dari fitur-fitur aslinya.Konsep & Rumus Matematika:Menghitung matriks kovarians $C$ dari data berpusat $X$:$$C = \frac{1}{n} X^T X$$Melakukan dekomposisi nilai eigen (eigenvalue decomposition) pada matriks kovarians untuk mencari nilai eigen ($\lambda$) dan vektor eigen ($v$):$$C v = \lambda v$$Vektor eigen dengan nilai eigen terbesar dipilih sebagai komponen utama pertama yang merepresentasikan arah varians data tertinggi.

## **3. Linear Discriminant Analysis (LDA)**
Berbeda dengan PCA yang tidak memperdulikan label kelas, LDA adalah teknik supervised dimensionality reduction. LDA tidak hanya berusaha memaksimalkan varians data, tetapi juga memaksimalkan pemisahan antar kelas (class separability). Hal ini dilakukan dengan meminimalkan varians di dalam kelas yang sama (within-class variance) sekaligus memaksimalkan jarak antara rata-rata kelas yang berbeda (between-class variance).

## **4. t-SNE untuk Visualisasi Data**
t-Distributed Stochastic Neighbor Embedding (t-SNE) adalah teknik non-linear yang sangat ampuh untuk visualisasi data berdimensi tinggi ke dalam ruang berdimensi rendah (2D atau 3D). t-SNE bekerja dengan mempertahankan kemiripan probabilitas antar tetangga dekat (local neighborhoods) di ruang asli dan ruang berdimensi rendah.

## **5. Contoh Program Penerapan 1: Mereduksi Dimensi Menggunakan PCA (Unsupervised)**
Contoh program ini memperlihatkan cara menggunakan PCA untuk mereduksi dataset Iris dari 4 fitur asli menjadi 2 komponen utama, serta melihat seberapa besar informasi (explained variance ratio) yang berhasil dipertahankan.

```python
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA

# 1. Memuat dataset Iris (memiliki 4 fitur: sepal length, sepal width, petal length, petal width)
iris = load_iris()
X, y = iris.data, iris.target

print(f"Bentuk/Dimensi data asli (X): {X.shape}")

# 2. Mereduksi dimensi dari 4 fitur menjadi 2 komponen utama (2D)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

print(f"Bentuk/Dimensi data setelah PCA: {X_pca.shape}")
print(f"Explained Variance Ratio (Informasi yang dipertahankan tiap komponen): {pca.explained_variance_ratio_}")
print(f"Total informasi yang dipertahankan: {sum(pca.explained_variance_ratio_) * 100:.2f}%")

print("\nContoh 5 data pertama setelah PCA:")
print(X_pca[:5])
```

## **6. Contoh Program Penerapan 2: Mereduksi Dimensi Menggunakan LDA (Supervised)**
Contoh program kedua ini menunjukkan penerapan LDA pada dataset yang sama. Karena LDA memanfaatkan informasi label kelas (y), teknik ini sangat ideal digunakan sebagai tahap pra-pemrosesan sebelum memasukkan data ke model klasifikasi.

```python
from sklearn.datasets import load_iris
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# 1. Memuat dataset Iris
iris = load_iris()
X, y = iris.data, iris.target

print(f"Bentuk/Dimensi data asli (X): {X.shape}")

# 2. Mereduksi dimensi menggunakan LDA dengan memanfaatkan label target (y)
# Catatan: Jumlah komponen maksimum untuk LDA adalah min(n_features, n_classes - 1)
lda = LinearDiscriminantAnalysis(n_components=2)
X_lda = lda.fit_transform(X, y)

print(f"Bentuk/Dimensi data setelah LDA: {X_lda.shape}")

print("\nContoh 5 data pertama setelah LDA:")
print(X_lda[:5])
```
